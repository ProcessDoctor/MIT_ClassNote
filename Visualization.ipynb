```python
#importing the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path= "/content/drive/MyDrive/Week 1  Foundations Python and Statistics/prog_book.csv"
data=pd.read_csv(path)
```

```python
data
```

```python
plt.hist(data['Price'])
#plt.hist(data['Price'], bins=30, color='blue', edgecolor='blue')
plt.xlabel('Price')
plt.ylabel('Frequency of Price')
plt.title('Histogram of Price')
```

```python
plt.figure(figsize=(12,6))
plt.xticks(rotation=45, ha='right')
sns.countplot(x=data['Type'], palette='Set2')
```

```python
#strip plot
plt.figure(figsize=(12,6))
sns.stripplot(x='Type', y='Rating', data=data, hue='Type')
```

```python
#boxplot
sns.boxplot(data['Rating'])
```

```python
#density plot shows the distribution
sns.kdeplot(data['Reviews'])
```
